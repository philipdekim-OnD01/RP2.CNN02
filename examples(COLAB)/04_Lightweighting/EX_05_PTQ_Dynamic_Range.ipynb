## Import Module

```python
import tensorflow as tf
import numpy as np

import tensorflow_model_optimization as tfmot
from tensorflow_model_optimization.python.core.keras.compat import keras

print(tf.__version__)
print(np.__version__)
```

## Load Dataset

```python
(train_images, train_labels), (test_images, test_labels) = keras.datasets.mnist.load_data()

train_images = (train_images / 255.0).astype(np.float32)
test_images = (test_images / 255.0).astype(np.float32)
```

## Load Baseline Model for MNIST (CNN)

```python
model = tf.keras.models.load_model('/content/drive/MyDrive/files/save/baseline_model.h5')
model.summary()
```

```python
_, baseline_model_accuracy = model.evaluate(
    test_images, test_labels, verbose=0)

print('Baseline test accuracy:', baseline_model_accuracy)
```

## Convert to TFLite model (Baseline model)

```python
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_baseline_model = converter.convert()
```

```python
litert_model_path = "/content/drive/MyDrive/files/save"
```

```python
import pathlib

tflite_models_dir = pathlib.Path(litert_model_path)
tflite_models_dir.mkdir(exist_ok=True, parents=True)

tflite_baseline_model_file = tflite_models_dir/"mnist_baseline_model.tflite"
size_baseline_model = tflite_baseline_model_file.write_bytes(tflite_baseline_model)
```

## Convert to TFLite model (Quantization model)

```python
converter.optimizations = [tf.lite.Optimize.DEFAULT] # default optimization strategy (enable post-training quantization)
tflite_ptq_dr_model = converter.convert()

tflite_ptq_dr_model_file = tflite_models_dir/"mnist_ptq_dr_model.tflite"
size_ptq_dr_model = tflite_ptq_dr_model_file.write_bytes(tflite_ptq_dr_model)
```

## File size 비교 : Baseline model vs PQT (DR) model

```python
print("Size of Baseline LiteRT Model file : {}".format(size_baseline_model))
print("Size of PTQ(DR) LiteRT Model file : {}".format(size_ptq_dr_model))
```

## interpreter 생성

```python
interpreter_base = tf.lite.Interpreter(model_path=str(tflite_baseline_model_file))
interpreter_base.allocate_tensors()
```

```python
input_details = interpreter_base.get_input_details()
output_details = interpreter_base.get_output_details()
```

## Define Interpreter for PTQ (dynamic range) Model

```python
interpreter_ptq_dr = tf.lite.Interpreter(model_path=str(tflite_ptq_dr_model_file))
interpreter_ptq_dr.allocate_tensors()
```

## input detail / output detail 확인

```python
input_details = interpreter_ptq_dr.get_input_details()
output_details = interpreter_ptq_dr.get_output_details()

input_dtype = input_details[0]['dtype']
output_dtype = output_details[0]['dtype']

input_shape = input_details[0]['shape']
output_shape = output_details[0]['shape']

print("input_dtype : {}".format(input_dtype))
print("output_dtype : {}".format(output_dtype))

print("input_shape : {}".format(input_shape))
print("output_shape : {}".format(output_shape))
```

## Run inference PQT(DR) model

```python
test_image = np.expand_dims(test_images[0], axis=0)

input_index = interpreter_ptq_dr.get_input_details()[0]["index"]
output_index = interpreter_ptq_dr.get_output_details()[0]["index"]

interpreter_ptq_dr.set_tensor(input_index, test_image)
interpreter_ptq_dr.invoke()
predictions = interpreter_ptq_dr.get_tensor(output_index)

print(predictions)
print(np.argmax(predictions))
print(test_labels[0])
```

```python
def evaluate_model(interpreter):
  input_index = interpreter.get_input_details()[0]["index"]
  output_index = interpreter.get_output_details()[0]["index"]

  prediction_digits = []
  for test_image in test_images:
    test_image = np.expand_dims(test_image, axis=0).astype(np.float32)
    interpreter.set_tensor(input_index, test_image)

    interpreter.invoke()

    output = interpreter.get_tensor(output_index)
    digit = np.argmax(output)
    prediction_digits.append(digit)

  accurate_count = 0
  for index in range(len(prediction_digits)):
    if prediction_digits[index] == test_labels[index]:
      accurate_count += 1
  accuracy = accurate_count * 1.0 / len(prediction_digits)

  return accuracy
```

```python
print(evaluate_model(interpreter_base))
print(evaluate_model(interpreter_ptq_dr))
```
